```python
import zipfile
import os

zip_path = "/content/data.zip"
extract_path = "/content/dataset"

if os.path.exists(zip_path):
    file_size = os.path.getsize(zip_path)
    print(f"File size of {zip_path}: {file_size} bytes")
    try:
        with open(zip_path, "rb") as f:
            header = f.read(4)
            print(f"File header bytes (hex): {header.hex()}")
            print(f"File header bytes (text): {header}")
            if header != b'PK\x03\x04':
                print("WARNING: The file does not start with the standard ZIP signature (PK\\x03\\x04).")
    except Exception as e:
        print(f"Could not read file header: {e}")
else:
    print(f"Error: {zip_path} does not exist.")

try:
    with zipfile.ZipFile(zip_path, "r") as zip_ref:
        zip_ref.extractall(extract_path)
    print("Dataset extracted successfully!")
    print(os.listdir(extract_path))
except zipfile.BadZipFile as e:
    print(f"\nExtraction failed with BadZipFile: {e}")
    print("Please verify that data.zip was fully uploaded and is not corrupted.")
```

```python
import os
from collections import Counter
from PIL import Image

dataset_path = "/content/dataset"

image_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".webp")
class_counts = Counter()
formats = Counter()
sizes = Counter()
total_images = 0

for root, dirs, files in os.walk(dataset_path):
    for file in files:
        if file.lower().endswith(image_extensions):
            path = os.path.join(root, file)

            try:
                with Image.open(path) as img:
                    total_images += 1
                    formats[img.format] += 1
                    sizes[img.size] += 1

                    relative_path = os.path.relpath(
                        root, dataset_path
                    )
                    class_name = relative_path.split(os.sep)[-1]
                    class_counts[class_name] += 1

            except Exception:
                pass

print("Total images:", total_images)
print("Image formats:", formats)
print("Image sizes:", sizes)
print("Images per folder:", class_counts)
```

```python
import os
from collections import Counter

image_extensions = {
    '.jpg', '.jpeg', '.png', '.bmp',
    '.webp', '.tif', '.tiff'
}

class_counts = Counter()

for split in ['train', 'training', 'test', 'testing', 'valid', 'validation']:
    split_path = os.path.join(extract_path, split)

    if os.path.isdir(split_path):
        for class_name in os.listdir(split_path):
            class_path = os.path.join(split_path, class_name)

            if os.path.isdir(class_path):
                count = sum(
                    1 for root, dirs, files in os.walk(class_path)
                    for f in files
                    if os.path.splitext(f)[1].lower() in image_extensions
                )
                if count > 0:
                    class_counts[f'{split}/{class_name}'] = count

if not class_counts:
    for root, dirs, files in os.walk(extract_path):
        for f in files:
            if os.path.splitext(f)[1].lower() in image_extensions:
                relative_path = os.path.relpath(root, extract_path)
                class_name = relative_path.split(os.sep)[-1]
                class_counts[class_name] += 1

print("Detected folders/classes:", len(class_counts))
print("\nImage count by detected folder:")
for name, count in sorted(class_counts.items()):
    print(f"{name}: {count}")
```

```python
# image size mode
from PIL import Image
from collections import Counter
import os

image_sizes = Counter()
image_modes = Counter()
channel_counts = Counter()
sample_count = 0
error_count = 0

for root, dirs, files in os.walk(extract_path):
    for file in files:
        if os.path.splitext(file)[1].lower() in image_extensions:
            path = os.path.join(root, file)

            try:
                with Image.open(path) as img:
                    image_sizes[img.size] += 1
                    image_modes[img.mode] += 1

                    channels = len(img.getbands())
                    channel_counts[channels] += 1
                    sample_count += 1

            except Exception:
                error_count += 1

print("Images inspected:", sample_count)
print("Unreadable images:", error_count)

print("\nImage dimensions (width, height):")
for size, count in image_sizes.most_common(15):
    print(f"{size}: {count} images")

print("\nImage modes:")
print(image_modes)

print("\nChannel counts:")
print(channel_counts)
```
